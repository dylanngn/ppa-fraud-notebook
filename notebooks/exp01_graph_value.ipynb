# Experiment 01: Graph Feature Value (RQ1)

**Research Question**: What novel relational indicators of fraud can be identified using graph-based analysis?

## Objective
Demonstrate that graph-derived features add significant value to fraud detection.

## Prerequisites
```bash
python -m src.experiments.exp01_graph_value
```

````python
import json
from pathlib import Path
from src.experiments.results import load_result, ExperimentID

# Load results
results = load_result(ExperimentID.EXP1_GRAPH_VALUE)
if results:
    print("Experiment 01 Results:")
    print("=" * 60)
    metrics = results.get('metrics', {})
    print(f"Full model (tabular + graph): {metrics.get('full_auc_pr', 'N/A'):.4f} AUC-PR")
    print(f"Tabular only:                 {metrics.get('tabular_only_auc_pr', 'N/A'):.4f} AUC-PR")
    print(f"Graph contribution:           +{metrics.get('graph_contribution', 'N/A'):.4f}")
    print(f"                              (+{metrics.get('graph_contribution_pct', 'N/A'):.1f}%)")
else:
    print("⚠️ Results not found. Run: python -m src.experiments.exp01_graph_value")
````

## Analysis

````python
if results:
    metrics = results.get('metrics', {})
    contribution = metrics.get('graph_contribution', 0)
    
    print("=" * 60)
    print("CONCLUSION")
    print("=" * 60)
    
    if contribution > 0.02:
        print("✅ Graph features provide SIGNIFICANT value")
        print(f"   Adding {metrics.get('n_graph_features', 17)} graph features improves AUC-PR by {contribution:.3f}")
    elif contribution > 0:
        print("⚠️ Graph features provide MARGINAL value")
    else:
        print("❌ Graph features provide NO measurable value")
````
